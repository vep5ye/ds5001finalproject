```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

from IPython.display import display, HTML
```

#### SENTIMENT VOCAB

```python
# load vocab
VOCAB = pd.read_csv("data/VOCAB.csv", index_col="term_str")
VOCAB
```

```python
# load sentiment lexicon
SALEX = pd.read_csv("lexicons/salex_nrc.csv").set_index('term_str')
SALEX.columns = [col.replace('nrc_','') for col in SALEX.columns]
SALEX
```

```python
# load LIB + CORPUS  as TOKENS
LIB = pd.read_csv("data/LIB.csv", index_col="mag_id").sort_index()

TOKENS = pd.read_csv(
    "data/CORPUS.csv",
    index_col=["mag_id", "page_num", "token_num"]).sort_index()

TOKENS = TOKENS.join(LIB)
TOKENS
```

```python
# join VOCAB + SALEX
VOCAB_SENT = VOCAB.join(SALEX, how="inner")
VOCAB_SENT
```

```python
# check shape
VOCAB_SENT.shape
```

```python
# save
VOCAB_SENT.to_csv("VOCAB_SENT.csv")
```

### SENTIMENT BOW

```python
BOW = pd.read_csv("BOW.csv", index_col=["mag_id", "term_str"])
VOCAB_SENT = pd.read_csv("VOCAB_SENT.csv", index_col="term_str")
```

```python
SENT_COLS = VOCAB_SENT.columns.drop("n", errors="ignore")

BOW_SENT = BOW.join(VOCAB_SENT[SENT_COLS], on="term_str", how="inner")
BOW_SENT
```

```python
BOW_SENT.shape
```

```python
BOW_SENT.to_csv("BOW_SENT.csv")
```

### SENTIMENT DOC_SENT

```python
BAG = ["mag_id"]

SENTIMENT_COLS = BOW_SENT.columns.drop(["n", "tf", "idf", "tfidf"], errors="ignore")

BAG_SENT = BOW_SENT[SENTIMENT_COLS].multiply(BOW_SENT["n"], axis=0)
BAG_SENT = BAG_SENT.groupby(BAG).sum()
BAG_SENT
```

```python
BAG_SENT.to_csv("BAG_SENT.csv")
```

### SENTIMENT PLOT

```python
# join sentiment w/ LIB metadata
LIB = pd.read_csv("data/LIB.csv", index_col="mag_id")

BAG_SENT_META = BAG_SENT.join(LIB[["month"]])
BAG_SENT_META
```

```python
# average sentiment by quarter
SENT_BY_MONTH = BAG_SENT_META.groupby("month").mean(numeric_only=True)
SENT_BY_MONTH
```

```python
SENT_BY_MONTH_PLOT = SENT_BY_MONTH.reset_index()

px.line(SENT_BY_MONTH_PLOT,
        x="month",
        y=SENT_BY_MONTH.columns,
        markers=True,
        title="Sentiment Over Time by Month",
        height=700)
```
