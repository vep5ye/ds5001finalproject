```python
import pandas as pd
import numpy as np

from sklearn.preprocessing import normalize
```

```python
TFIDF = pd.read_csv("TFIDF.csv", index_col="mag_id")
TFIDF
```

```python
# Reduce features by keeping words with total TFIDF above 0
SIG_WORDS = TFIDF.sum().sort_values(ascending=False)
SIG_WORDS = SIG_WORDS[SIG_WORDS > 0].index
SIG_WORDS
```

```python
TFIDF_REDUCED = TFIDF[SIG_WORDS]
TFIDF_REDUCED
```

```python
# L2 normalize
TFIDF_L2 = pd.DataFrame(
    normalize(TFIDF_REDUCED, norm='l2'),
    index=TFIDF_REDUCED.index,
    columns=TFIDF_REDUCED.columns)

TFIDF_L2
```

```python
TFIDF_L2.to_csv('TFIDF_L2.csv')
```
