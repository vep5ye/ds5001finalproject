```python
import pandas as pd
import numpy as np
import plotly.express as px

from sklearn.decomposition import PCA
```

```python
TFIDF_L2 = pd.read_csv('TFIDF_L2.csv', index_col='mag_id')
TFIDF_L2
```

```python
pca_engine = PCA(n_components=10, random_state=42)

DCM = pd.DataFrame(pca_engine.fit_transform(TFIDF_L2),
                   index=TFIDF_L2.index)

DCM.columns = [f"PC{i}" for i in range(10)]
DCM
```

```python
COMPS = pd.DataFrame(pca_engine.components_.T * np.sqrt(pca_engine.explained_variance_), index=TFIDF_L2.columns)

COMPS.columns = [f"PC{i}" for i in range(10)]
COMPS.index.name = "term_str"

COMPS.T
```

```python
PC0 = COMPS["PC0"].sort_values(ascending=False).head(5).to_frame("PC0")
PC0
```

```python
PC1 = COMPS["PC1"].sort_values().head(5).to_frame("PC1")
PC1
```

```python
# data for plotting
LIB = pd.read_csv('data/LIB.csv', index_col=['mag_id'])
LIB
```

```python
# Plot 1: documents in PCA space
DCM_META = DCM.join(LIB[['quarter']])

px.scatter(DCM_META.reset_index(),
           x="PC0",
           y="PC1",
           color="quarter",
           hover_name="mag_id",
           title="PCA Document Scatterplot by Quarter",
           height=400,
           width=600)
```

```python
# loadings scatterplot 
LOADINGS = COMPS[["PC0", "PC1"]].reset_index()

px.scatter(LOADINGS,
           x="PC0",
           y="PC1",
           text="term_str",
           title="PCA Loadings Scatterplot",
           height=500,
           width=700)
```

### PCA VISUALIZATION 1
- The first component shows a clear split in vocabulary, separating documents based on different themes or types of language.

```python
px.scatter(DCM_META.reset_index(),
           x="PC2",
           y="PC3",
           color="quarter",
           hover_name="mag_id",
           title="PCA Document Scatterplot by Quarter: PC2 vs PC3",
           height=500,
           width=700)
```

```python
LOADINGS_23 = COMPS[["PC2", "PC3"]].reset_index()

px.scatter(LOADINGS_23,
           x="PC2",
           y="PC3",
           hover_name="term_str",
           title="PCA Loadings Scatterplot: PC2 vs PC3",
           height=500,
           width=700)
```

### PCA VISUALIZATION 2
- The second component shows a more subtle split in vocabulary. It’s not as strong as the first, but it still separates documents based on smaller differences in language and themes.
