```python
import pandas as pd
import numpy as np
```

```python
CORPUS = pd.read_csv("data/CORPUS.csv", index_col=['mag_id','page_num','token_num'])
CORPUS
```

```python
BAG = ['mag_id']
BAG
```

```python
BOW = CORPUS.groupby(BAG + ["term_str"]).term_str.count().to_frame("n")
BOW
```

```python
BOW.shape[0]
```

```python
D = BOW.groupby(BAG).n.sum()
BOW["tf"] = BOW["n"] / D

N = BOW.groupby(BAG).ngroups
DF = BOW.groupby("term_str")["n"].count()
IDF = np.log2(N / DF)

BOW["idf"] = BOW.index.get_level_values("term_str").map(IDF)
BOW["tfidf"] = BOW["tf"] * BOW["idf"]

BOW
```

```python
BOW.to_csv('BOW.csv')
```
